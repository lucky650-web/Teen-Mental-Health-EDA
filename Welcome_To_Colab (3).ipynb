```
df=df.drop_duplicates()
```

```
plt.figure(figsize=(8,5))
sns.histplot(df['age'],bins=10,kde=True)
plt.title('Age Distribution')
plt.show()
```

```
plt.figure(figsize=(8,4))
sns.histplot(df['daily_social_media_hours'],bins=15)
plt.title('Daily Social Media Hours')
plt.show()
```

```
plt.figure(figsize=(8,4))
sns.countplot(x='platform_usage',data=df)
plt.title('platform usage')
plt.xticks(rotation=45)
plt.show()
```

```
plt.figure(figsize=(8,4))
sns.histplot(df['sleep_hours'],bins=15,kde=True)
plt.title('Sleep Hours')
plt.show()
```

```
plt.figure(figsize=(8,4))
sns.countplot(x='stress_level',data=df)
plt.title('stress_level')
plt.show()
```

```
plt.figure(figsize=(8,4))
sns.countplot(x='anxiety_level',data=df)
plt.title('anxiety_level')
plt.show()
```

```
plt.figure(figsize=(6,4))
sns.countplot(x='depression_label',data=df)
plt.title('depression_label')
plt.show()
```

```
plt.figure(figsize=(8,5))
sns.scatterplot(x='daily_social_media_hours',y='stress_level', data=df)
plt.title("Social Media Usage vs Stress Level")
plt.show()
```

```
plt.figure(figsize=(8,5))
sns.scatterplot( x='sleep_hours',y='anxiety_level',data=df)
plt.title("Sleep Hours vs Anxiety")
plt.show()
```

```
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])
```

```
plt.figure(figsize=(12,8))
sns.heatmap( df.corr(),annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```
sns.pairplot(
    df[['sleep_hours',
        'stress_level',
        'anxiety_level',
        'addiction_level']]
)
plt.show()
```

```
fig, axes = plt.subplots(2,2, figsize=(14,10))
sns.countplot(x='gender', data=df, ax=axes[0,0])
axes[0,0].set_title("Gender")
sns.countplot(x='stress_level', data=df, ax=axes[0,1])
axes[0,1].set_title("Stress")
sns.histplot(df['sleep_hours'], ax=axes[1,0])
axes[1,0].set_title("Sleep Hours")
sns.countplot(x='depression_label', data=df, ax=axes[1,1])
axes[1,1].set_title("Depression")
plt.tight_layout()
plt.show()
```
